## Background and Prior Work

**Jungian Theory and Myers-Briggs Typology Indicator**<br>

When it comes to MBTI, they are considered to be the most known forms of typology. BIG 5 or OCEAN is also considered to be noteworthy and incredibly useful. However, it does not seem to be as popular as the prior two considered. I'd like to preface this by explaining exactly what MBTI is and how it differs from the original Jungian functions. 

Carl Jung created 8 Cognitive Functions: Extraverted/Introverted Intuition, Sensing, Feelings and Thinking (Ne,Ni,Se,Si,Fe,Fi,Te,Ti). Years later, Myers and Briggs took his knowledge and ideas and created 16 personalities from the original idea. This was later known as MBTI types. MBTI types being made up of 8 phrases: Extravert/Introvert, Sensing/Intuition, Feeling/Thinking, and Judging/Perceiving. The 8 letters became 4 choices, and you are labeled into one of 16 boxes made up of the different letter scheme. 

According to TheMyersBriggs.com, The Myers-Briggs Type Indicator® (MBTI®) assessment is one of the world’s most popular personality tools—because it works. Used by more than 88 percent of Fortune 500 companies in 115 countries, and available in 29 languages, it has become the go-to framework for people development globally. With more than 70 years of science-based, research-based insight, the MBTI assessment is a robust tool for self-awareness and improvement. It provides positive language for understanding and valuing individual differences. With practical insight that’s easy to understand and implement, the MBTI assessment has helped thousands of organizations and millions of people around the world improve how they communicate, learn, and work<sup>[1](#themyersbriggs)</sup>.

Several other studies have looked further into the idea of patients with several mental health disorders and their correspondent typology type. One study in specifics looked further into patients with suicidal tendencies and thoughts<sup>[2](#suicidal)</sup>. In addition, there has been more reasearch into specific mental health disorders and the connection to different letters in MBTI<sup>[3](#unipolar)</sup>.

Although there are many hypotheses and assumptions in regards to typology and mental health disorders, there is no clear evidentiary support. While many psychologists, including Carl Jung himself, state that there are no strong connection between the two, it is something that could be considered useful to look further into<sup>[4](#darkside)</sup>

## OLS Attempt (Linear Regression):

```python
mbti_cols = [col for col in grouped_df_mbti.columns if col != "Country"]
analysis_df = grouped_df_mbti.merge(grouped_df_mental, on="Country")
analysis_df[mbti_cols] = analysis_df[mbti_cols].div(analysis_df[mbti_cols].sum(axis=1), axis=0)

mental_cols = ["Schizophrenia", "Depression", "Anxiety", "Bipolar", "Eating Disorders"]

# Collect results
results = []

for mbti in mbti_cols:
    for disorder in mental_cols:
        X = sm.add_constant(analysis_df[[mbti]])  # single MBTI type
        y = analysis_df[disorder]
        
        model = sm.OLS(y, X).fit()
        
        results.append({
            "MBTI": mbti,
            "Disorder": disorder,
            "R²": model.rsquared,
            "Adj. R²": model.rsquared_adj,
            "Coef": model.params[mbti],
            "p-value": model.pvalues[mbti]
        })

# Convert to DataFrame
summary_table = pd.DataFrame(results)

print(summary_table.head(20))
```

```python
coef_matrix = summary_table.pivot(index="MBTI", columns="Disorder", values="Coef")

# Pivot for p-values
pval_matrix = summary_table.pivot(index="MBTI", columns="Disorder", values="p-value")

# Pivot for R²
r2_matrix = summary_table.pivot(index="MBTI", columns="Disorder", values="R²")

print(coef_matrix)
print(pval_matrix)
print(r2_matrix)
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(coef_matrix, annot=True, cmap="coolwarm", center=0)
plt.title("OLS Coefficients: MBTI vs Mental Disorders")
plt.show()

plt.figure(figsize=(10, 6))
sns.heatmap(pval_matrix, annot=True, cmap="viridis_r")
plt.title("P-values: MBTI vs Mental Disorders")
plt.show()

plt.figure(figsize=(10, 6))
sns.heatmap(r2_matrix, annot=True, cmap="Blues")
plt.title("R² Values: MBTI vs Mental Disorders")
plt.show()
```

```python
sig_mask = pval_matrix >= 0.05
plt.figure(figsize=(10, 6))
sns.heatmap(coef_matrix, annot=True, cmap="coolwarm", center=0, mask=sig_mask)
plt.title("Significant OLS Coefficients (p < 0.05)")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(pval_matrix < 0.05, annot=True, cmap="Greens", cbar=False)
plt.title("Significance Map (True = p < 0.05)")
plt.show()
```

### WELL WELL WELL

While some of these do look like true, and possible, correlations. I want to double check by doing some stricter analyses!

```python
summary_adj = summary_table.copy()

# Bonferroni (very strict)
_, p_bonf, _, _ = multipletests(summary_adj["p-value"], method="bonferroni")

# FDR / Benjamini-Hochberg (less strict, controls false discovery rate)
_, p_fdr, _, _ = multipletests(summary_adj["p-value"], method="fdr_bh")

summary_adj["p-bonf"] = p_bonf
summary_adj["p-fdr"] = p_fdr

summary_adj = summary_table.copy()

# Bonferroni (very strict)
_, p_bonf, _, _ = multipletests(summary_adj["p-value"], method="bonferroni")

# FDR / Benjamini-Hochberg (less strict, controls false discovery rate)
_, p_fdr, _, _ = multipletests(summary_adj["p-value"], method="fdr_bh")

# Add back into table
summary_adj["p-bonf"] = p_bonf
summary_adj["p-fdr"] = p_fdr
```
